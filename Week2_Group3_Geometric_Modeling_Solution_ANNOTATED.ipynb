# PART I: General Geometric Tasks

```python
# Define three given 2D points
# Each point is represented as a NumPy array (x, y)
# Expected output: a dictionary containing 2D coordinate pairs

points_2d = {
    "P1": np.array([2, 3]),     # Point in the 1st quadrant
    "P2": np.array([-3, 1]),    # Point in the 2nd quadrant
    "P3": np.array([1, -2])     # Point in the 4th quadrant
}

# Display the dictionary
# Expected output: printed coordinates for P1, P2, and P3
points_2d
```

```python
# Plot the defined 2D points on a Cartesian plane
# Expected output:
# - A 2D scatter plot
# - Points positioned according to their (x, y) coordinates
# - X-axis and Y-axis clearly shown
# - Each point labeled (P1, P2, P3)

import matplotlib.pyplot as plt

# Loop through each point in the dictionary
for name, p in points_2d.items():
    # Plot the point
    plt.scatter(p[0], p[1])

    # Add a text label slightly offset from the point
    plt.text(p[0] + 0.1, p[1] + 0.1, name)

# Draw the coordinate axes
# These help visualize quadrants
plt.axhline(0, color='black', linewidth=0.8)  # X-axis
plt.axvline(0, color='black', linewidth=0.8)  # Y-axis

# Label the axes
plt.xlabel("X-axis")
plt.ylabel("Y-axis")

# Add title
plt.title("2D Cartesian Coordinate Plot")

# Add grid for better readability
plt.grid(True)

# Display the plot
# Expected: A clear 2D Cartesian plot with labeled points
plt.show()
```

## Part A: Coordinate Systems and Point Representation

```python

# Define three given 3D points
# Expected output: dictionary with 3D coordinates

points_3d = {
    "P1": np.array([2, 3, 4]),
    "P2": np.array([-3, 1, 2]),
    "P3": np.array([1, -2, -3])
}

points_3d
```

```python

# Function to identify octant
# Expected output: name of the octant for each point

def find_octant(p):
    x, y, z = p
    if x>0 and y>0 and z>0: return "1st Octant"
    if x<0 and y>0 and z>0: return "2nd Octant"
    if x<0 and y<0 and z>0: return "3rd Octant"
    if x>0 and y<0 and z>0: return "4th Octant"
    if x>0 and y>0 and z<0: return "5th Octant"
    if x<0 and y>0 and z<0: return "6th Octant"
    if x<0 and y<0 and z<0: return "7th Octant"
    if x>0 and y<0 and z<0: return "8th Octant"

for name, point in points_3d.items():
    print(name, ":", find_octant(point))
```

```python
# Function to identify the octant of a 3D point
# Expected output: octant name based on signs of x, y, z

def find_octant(p):
    x, y, z = p
    if x > 0 and y > 0 and z > 0:
        return "1st Octant"
    elif x < 0 and y > 0 and z > 0:
        return "2nd Octant"
    elif x < 0 and y < 0 and z > 0:
        return "3rd Octant"
    elif x > 0 and y < 0 and z > 0:
        return "4th Octant"
    elif x > 0 and y > 0 and z < 0:
        return "5th Octant"
    elif x < 0 and y > 0 and z < 0:
        return "6th Octant"
    elif x < 0 and y < 0 and z < 0:
        return "7th Octant"
    elif x > 0 and y < 0 and z < 0:
        return "8th Octant"
    else:
        return "On a coordinate plane"

# Identify the octant for each point
# Expected output: octant classification for each point
for name, point in points_3d.items():
    print(name, "lies in the", find_octant(point))
```

```python
# Plot the 3D points using Matplotlib
# Expected output:
# - A 3D scatter plot
# - Each point positioned correctly in space
# - Axes labeled X, Y, Z
# - Points labeled as P1, P2, P3
# {'P1': array([2, 3, 4]), 'P2': array([-3,  1,  2]), 'P3': array([ 1, -2, -3])}


import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Create a 3D plotting area
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot each point
for name, p in points_3d.items():
    ax.scatter(p[0], p[1], p[2])
    ax.text(p[0] + 0.1, p[1] + 0.1, p[2] + 0.1, name)

# Label the axes
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.set_zlabel("Z-axis")

# Add title
plt.title("3D Cartesian Coordinate Plot")

# Display the plot
plt.show()
```

## Part B: Points, Vectors & Distance

```python

# Define points A and B
# Expected output: vectors for A and B

A = np.array([2, -1, 1])
B = np.array([1, -1, 3])

# Vector AB
# Expected: direction vector from A to B
AB = B - A

# Distance |AB|
# Expected: positive scalar value
distance = np.linalg.norm(AB)

# Midpoint of AB
# Expected: coordinate midpoint
midpoint = (A + B) / 2

AB, distance, midpoint
```

```python
# Visualize points A, B, the vector AB, and the midpoint in 3D
# Expected output:
# - A 3D plot
# - Points A and B shown as dots
# - Vector AB shown as an arrow from A to B
# - Midpoint shown as a different colored point

import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

# Create 3D plotting area
fig = plt.figure()
ax = fig.add_subplot(111, projection='3d')

# Plot point A
ax.scatter(A[0], A[1], A[2], color='blue', s=60)
ax.text(A[0], A[1], A[2], 'A', color='blue')

# Plot point B
ax.scatter(B[0], B[1], B[2], color='green', s=60)
ax.text(B[0], B[1], B[2], 'B', color='green')

# Plot vector AB using quiver (arrow)
# The arrow starts at A and points in direction AB
ax.quiver(
    A[0], A[1], A[2],      # Starting point
    AB[0], AB[1], AB[2],   # Direction vector
    color='red', linewidth=2
)

# Plot midpoint
ax.scatter(midpoint[0], midpoint[1], midpoint[2], color='purple', s=80)
ax.text(midpoint[0], midpoint[1], midpoint[2], 'Midpoint', color='purple')

# Label axes
ax.set_xlabel("X-axis")
ax.set_ylabel("Y-axis")
ax.set_zlabel("Z-axis")

# Set title
plt.title("3D Visualization of Points A, B, Vector AB, and Midpoint")

# Show plot
plt.show()
```

## Part C: Vector Operations

```python

# Define two vectors
v1 = np.array([3, 2, -1])
v2 = np.array([1, -2, 4])

# Vector addition
# Expected: new vector
addition = v1 + v2

# Dot product
# Expected: scalar
dot_product = np.dot(v1, v2)

# Cross product
# Expected: perpendicular vector
cross_product = np.cross(v1, v2)

addition, dot_product, cross_product
```

# PART II: Real Dataset-Based Tasks – Group 3

### Load Sample GPS Trajectory Data

```python

# Sample GPS data (replace with real Geolife file)
# Expected output: table of latitude and longitude

df = pd.DataFrame({
    "latitude": [39.984, 39.985, 39.986, 39.987, 39.989],
    "longitude": [116.318, 116.319, 116.321, 116.323, 116.325]
})

df
```

### Compute Displacement Vectors

```python

# Compute displacement vectors
# Expected: list of vectors showing movement

vectors = []

for i in range(len(df)-1):
    p1 = np.array([df.loc[i,'longitude'], df.loc[i,'latitude']])
    p2 = np.array([df.loc[i+1,'longitude'], df.loc[i+1,'latitude']])
    vectors.append(p2 - p1)

vectors
```

### Vector Magnitudes and Dot Products

```python

# Compute magnitudes
# Expected: list of positive values
magnitudes = [np.linalg.norm(v) for v in vectors]

# Compute dot products
# Expected: values indicating direction similarity
dot_products = [np.dot(vectors[i], vectors[i+1]) for i in range(len(vectors)-1)]

magnitudes, dot_products
```

### Plot GPS Trajectory

```python

# Plot trajectory
# Expected: 2D path of movement

plt.plot(df['longitude'], df['latitude'], marker='o')
plt.xlabel("Longitude")
plt.ylabel("Latitude")
plt.title("GPS Trajectory")
plt.grid(True)
plt.show()
```


## Interpretation
- Magnitude shows movement distance
- Dot product shows direction change
- Useful for motion analysis and modeling
